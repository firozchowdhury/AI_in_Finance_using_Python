```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 1000 # Number of throws
N = 2 # Number of dice

sum = np.random.randint(1, 7, size=(n, N)).sum(axis=1) # numpy random formula
mu = sum.mean() # μ (mu) usually represents the mean (average)
sigma = sum.std() # σ (sigma) usually represents the standard deviation
counts, bins, bars = plt.hist(sum, bins=range(N, 6*N+2),edgecolor='black', align='left')
plt.ylabel("Frequency")
plt.title(f"Distribution of sums of {N:d} dice (mean = {mu:.2f}, std ={sigma:.2f} {n:d} throws ")
plt.show()

probs = 100*counts/n
```

The histogram shows the frequency distribution of the sums obtained from throwing two fair dice 1000 times. The distribution is symmetric and centered near 7. The sum 7 occurs most frequently, while 2 and 12 occur least frequently. The mean is approximately 7, and the standard deviation shows that most sums are close to 7. The simulation agrees with the theoretical probabilities of rolling two dice.

```python
import numpy as np
import matplotlib.pyplot as plt

# Experiment 1
n1 = 1000
sums1 = np.random.randint(1, 7, size=(n1, 2)).sum(axis=1)
mu1 = sums1.mean()

# Experiment 2
n2 = 100000
sums2 = np.random.randint(1, 7, size=(n2, 2)).sum(axis=1)
mu2 = sums2.mean()

# Two plots in one row
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Left plot
ax[0].hist(sums1, bins=range(2, 14),
           edgecolor='black', align='left')
ax[0].set_title(f"1000 Throws\nMean = {mu1:.4f}")

# Right plot
ax[1].hist(sums2, bins=range(2, 14),
           edgecolor='black', align='left')
ax[1].set_title(f"100000 Throws\nMean = {mu2:.4f}")

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import norm

# Parameters for the normal distribution
mean = 7
std = 2.37

# Generate x-values
x = np.linspace(mean - 4, mean + 4, 500)

# Compute the normal PDF
y = norm.pdf(x, mean, std)

# Plot
plt.plot(x, y, label=f"Normal(μ={mean}, σ={std})")
plt.title("Bell Curve (Normal Distribution)")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.grid(True)
plt.legend()
plt.show()
```
