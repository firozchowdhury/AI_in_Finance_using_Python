```python
#all
import pandas as pd

df = pd.read_csv("sales_data.csv")

# Correlation matrix
print(df.corr(numeric_only=True))

# Total sales amount grouped by region
print(df.groupby("Region")[["Sales_Amount"]].sum())

# Maximum sales amount grouped by sales representative
print(df.groupby("Sales_Rep")[["Sales_Amount"]].max())

# Mean unit price grouped by product category
print(df.groupby("Product_Category")[["Unit_Price"]].mean())
```
