```markdown
# Learning Diary – AI in Finance using Python, 3 ECTS

**Student name:** [Your Name]  
**Student number:** [Your Metropolia student number]  
**Email:** [Your Metropolia email]  
**Course platform:** Viope – AI in Finance using Python, 3 ECTS[page:21]  
**Diary period:** [Start date] – [End date]  

---

## 1. Introduction

In this learning diary, I reflect on what I have learned from the course **AI in Finance using Python** and especially from the **open exercises**.[page:21]  
My background is in accounting/bookkeeping, and I am interested in applying Python and AI methods to financial and SME accounting data.

Key aims of my learning in this course:
- Understand basic machine learning concepts in finance.
- Practice Python-based data analysis on financial datasets.
- Connect course topics to my future thesis on fraud and risk analytics.

---

## 2. Course structure and my progress

### 2.1 Main course parts

The course consists of the following main sections in Viope:[page:1]

1. Preface  
2. Python environment  
3. Data acquisition etc.  
4. Timeseries analysis  
5. Sentiment analysis  
6. Risk management  
7. Fraud detection  
8. Portfolio optimization  

### 2.2 My progress and study style

- I have completed approximately [X]% of all assignments (target: at least 80%).[page:21]  
- I mainly studied by:
  - Reading the theory in each section.  
  - Running and modifying the Python examples.  
  - Answering the open exercises with explanations and code.

Reflection:
- What worked well in my study routine?
- What was challenging (e.g. model tuning, understanding risk measures)?
- How did I manage time and motivation?

---

## 3. Key learnings from open exercises

In line with Metropolia’s instructions, I focus here on **what I learned from the open exercises** in each topic.[page:21]

### 3.1 Python environment and data acquisition

- I learned how to set up a Python environment suitable for financial data analysis (libraries, Jupyter, etc.).[page:1]  
- The exercises helped me practise loading CSV/Excel data, handling dates, and cleaning financial datasets.

Reflections:
- How this improved my confidence in using Python for accounting/finance.  
- Any specific problems I solved (e.g., missing values, inconsistent formats).

### 3.2 Timeseries analysis

- I experimented with time series plots and basic analysis (trends, seasonality) on financial data.[page:1]  
- The open exercises showed how to calculate and visualise rolling averages, returns, or other time-based metrics.

Reflections:
- How timeseries thinking can be applied to cash flows or SME accounting data.  
- Any insights about volatility, trends, or unusual patterns.


অবশ্যই। তোমার Viope কোর্সের ভিত্তিতে একটি LSTM Stock Forecasting Roadmap দিচ্ছি। এটা Thesis, Project বা Portfolio—সব ক্ষেত্রেই ব্যবহার করতে পারবে।

LSTM Forecasting Project Workflow
1. Collect Data
        ↓
2. Clean Data
        ↓
3. Feature Engineering
        ↓
4. Scale Data
        ↓
5. Create Sequences
        ↓
6. Train-Test Split
        ↓
7. Build LSTM
        ↓
8. Train Model
        ↓
9. Predict
        ↓
10. Evaluate
        ↓
11. Visualize
        ↓
12. Deploy Dashboard

Step 1: Collect Data
import yfinance as yf

df = yf.download(
    "AAPL",
    start="2015-01-01",
    end="2025-01-01"
)

df.reset_index(inplace=True)


Output:

Date
Open
High
Low
Close
Volume

Step 2: Data Cleaning

Check missing values:

print(df.isnull().sum())


Fill if necessary:

df.fillna(method="ffill", inplace=True)

Step 3: Feature Engineering
Moving Average
df['MA20'] = df['Close'].rolling(20).mean()

df['MA50'] = df['Close'].rolling(50).mean()

Daily Return
df['Return'] = df['Close'].pct_change()

Volume Change
df['Volume_Change'] = df['Volume'].pct_change()


Remove NaNs:

df.dropna(inplace=True)

Step 4: Select Features

Example:

features = [
    'Open',
    'High',
    'Low',
    'Close',
    'Volume',
    'MA20',
    'MA50'
]

df = df[features]

Step 5: Scale Data
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)


Convert back:

scaled_data = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

Step 6: Create Sequences

Create a 20-day window:

def sampler(df, window):

    X = []
    y = []

    for i in range(len(df)-window):

        X.append(
            df.iloc[i:i+window].values
        )

        y.append(
            df['Close'].iloc[i+window]
        )

    return np.array(X), np.array(y)


Run:

X, y = sampler(
    scaled_data,
    20
)

Step 7: Split Train/Test
split = 0.8

train_size = int(
    len(X)*split
)

X_train = X[:train_size]
X_test = X[train_size:]

y_train = y[:train_size]
y_test = y[train_size:]

Step 8: Build LSTM Model
import keras

model = keras.Sequential()


LSTM layer:

model.add(
    keras.layers.LSTM(
        100,
        input_shape=(
            X_train.shape[1],
            X_train.shape[2]
        )
    )
)


Dropout:

model.add(
    keras.layers.Dropout(0.2)
)


Output layer:

model.add(
    keras.layers.Dense(1)
)


Compile:

model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

Step 9: Train Model
history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32
)

Step 10: Predict
predictions = model.predict(
    X_test
)

Step 11: Evaluate
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)


MSE:

mse = mean_squared_error(
    y_test,
    predictions
)


MAE:

mae = mean_absolute_error(
    y_test,
    predictions
)


R²:

r2 = r2_score(
    y_test,
    predictions
)


Print:

print(mse)
print(mae)
print(r2)

Step 12: Actual vs Predicted Graph
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.plot(
    y_test,
    label='Actual'
)

plt.plot(
    predictions,
    label='Predicted'
)

plt.legend()

plt.show()

Step 13: Next-Day Forecast

Take the latest 20 days:

latest = scaled_data.iloc[-20:]

latest = latest.values.reshape(
    1,
    20,
    scaled_data.shape[1]
)


Predict:

forecast = model.predict(latest)


Output:

Tomorrow Close Price Prediction

Step 14: Buy/Hold/Sell Logic
current = df['Close'].iloc[-1]

predicted = forecast[0][0]

if predicted > current * 1.02:
    signal = "BUY"

elif predicted < current * 0.98:
    signal = "SELL"

else:
    signal = "HOLD"

Step 15: Dashboard (Future Thesis Step)

Use:

Streamlit


Pages:

Stock Selection
Price Chart
Technical Indicators
LSTM Prediction
Buy/Hold/Sell Signal
Model Performance

Final Thesis Structure
Phase 1
Yahoo Finance Data

Phase 2
Technical Indicators

Phase 3
Scaling

Phase 4
LSTM

Phase 5
Prediction

Phase 6
MSE / MAE / R²

Phase 7
Dashboard


এটাই একটি সম্পূর্ণ end-to-end AI Stock Forecasting System-এর roadmap, এবং HAMK thesis-এর জন্য যথেষ্ট শক্তিশালী foundation।

### 3.3 Sentiment analysis

- I learned how sentiment analysis can be applied to financial news or text data.[page:1]  
- The exercises introduced basic text processing and sentiment scoring in Python.

Reflections:
- How sentiment could affect portfolio decisions or risk perception.  
- Whether sentiment analysis might be useful for my own future projects.

### 3.4 Risk management

- I gained an understanding of common risk measures and how to compute them in Python.[page:1]  
- The exercises showed how to look at risk from a portfolio or financial perspective.

Reflections:
- How these risk concepts relate to accounting (e.g. cash-flow risk, credit risk).  
- Which measures I found most intuitive or useful.

### 3.5 Fraud detection

- I learned basic ideas for detecting unusual or suspicious financial transactions using data analysis.[page:1]  
- The open exercises encouraged me to think about anomaly detection and rule-based approaches.

Reflections:
- How this connects directly to my interest in fraud detection in SME accounting.  
- Specific patterns or methods that I would like to explore further.

### 3.6 Portfolio optimization

- I practised simple portfolio optimization and saw how risk/return trade-offs can be analysed.[page:1]  
- The exercises made me more aware of diversification and constraints.

Reflections:
- How portfolio thinking might be useful for small investors or businesses.  
- Whether these methods could be used together with my accounting knowledge.

---

## 4. Applying the course to my own goals

### 4.1 Link to my accounting and career plans

- The course strengthened my ability to use Python and AI methods on financial and accounting data.[page:1]  
- I now see practical ways to apply data analysis to bookkeeping, fraud detection, and risk management.

Examples:
- Analysing SME transaction data for anomalies.  
- Building small dashboards for cash-flow or expense monitoring.  
- Supporting internal control and financial decision-making.

### 4.2 Link to my thesis idea

My working thesis idea is:

> **Python-based fraud and risk analytics for SME accounting data.**

This course supports that topic because:
- It introduced timeseries analysis and risk concepts that can be applied to accounting data.[page:1]  
- It provided initial ideas for fraud detection and anomaly analysis using Python.[page:1]

In the future, I plan to:
- Use similar Python workflows as in the course for my own datasets.  
- Combine rule-based accounting checks with data-driven anomaly detection.

---

## 5. Reflection on AI in finance and limitations

- The course emphasised that financial AI models have limitations, especially in predicting markets and rare events.[page:1]  
- This made me more aware of the risks of over-trusting models and the need for human judgement.

Reflections:
- Why education and transparency are important when using AI in finance.  
- How I will balance model results with professional accounting judgement.

---

## 6. Overall evaluation of my learning

### 6.1 What I gained

- Better understanding of Python in financial contexts.  
- Practical experience with datasets and open exercises.  
- A clearer view of how AI and machine learning fit into real financial work.[page:1]

### 6.2 What I still need to improve

- Deeper understanding of specific algorithms (e.g., anomaly detection, classification).  
- More practice with larger and more complex datasets.  
- Stronger connection between theory and everyday accounting tasks.

---

## 7. Plans for further study

- Continue practising Python with real or simulated SME accounting data.  
- Study more detailed articles and textbooks on fraud detection and risk management.  
- Prepare a thesis plan that builds on this course, including a clear methodology and datasets.

---

*Date:* [Write date here]  
*Signature (if needed):* [Your Name]
```
