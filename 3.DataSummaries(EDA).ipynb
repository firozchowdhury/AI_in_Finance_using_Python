### 1. MEAN (average)
The mean formula, step by step

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i = \frac{1}{n}(x_1 + x_2 + \cdots + x_n)
$$




$$
\bar{x}
$$  
means the mean or average.
$$
\frac{1}{n}\sum_{i=1}^{n}x_i
$$ 
means add all values from 1 to n, where n is numbers, i=1 to i=n

### 2. Median
2 average is grouped, one is low and another is high, So, Median is the average of those 2 group's average value.

### 3. Quartiles
Quartiles also split the data into parts.

Q1 is the first quartile, where 25% of the numbers are below it.

Q2 is the second quartile, and that is the same as the median.

Q3 is the third quartile, where 75% of the numbers are below it, or 25% are above it.

So quartiles help you see where the data is concentrated.

### 4. Variance
Variance tells you how far the numbers are from the mean on average.

If the variance is large, the numbers are more spread out
So, If the numbers are close together, the variance is small and If the numbers are far apart, the variance is large

$$
\mathrm{Var}(x)=\frac{1}{n}\sum_{i=1}^{n}(x_i-\bar{x})^2
$$
(how far one value is from the mean)
In simple steps
    Find the average.

    Subtract the average from each number.

    Square each result.

    Add them all.

    Divide by how many numbers there are.

So variance is basically the average squared distance from the mean


### 5. standard deviation
Variance tells you how spread out the numbers are, but its value is in squared units
Taking the square root turns it back into the original units, so it is easier to understand.
$$
\sigma = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i-\bar{x})^2}
$$
